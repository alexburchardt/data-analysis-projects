```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("fitness_data.csv")
```

```python
plt.plot(df["steps"])
plt.title("Raw Steps Table")
plt.ylabel("Steps")
plt.show()
```

```python
plt.plot(df["date"],df["steps"])
plt.title("Number of Steps Per Day")
plt.xlabel("Date")
plt.ylabel("Number of Steps")
```

```python
plt.figure(figsize=(15,5))
plt.plot(df["date"], df["resting_hr"], color="rebeccapurple")
plt.title("Resting Heart Rate Over Time", fontsize=22)
plt.ylabel("Heart Rate")
plt.xlabel("Day")
```

```python
plt.figure(figsize=(6,6))
plt.scatter(df["steps"],df["calories_burned"],alpha=0.6,color="rebeccapurple")
plt.title("Steps vs Calories Burned", fontsize=22)
plt.xlabel("Steps", fontsize=10)
plt.ylabel("Calories Burned", fontsize=10)
plt.show()
```

```python
df.head()
```

```python
plt.figure(figsize=(7,7))
colors = {"Low": "lightblue", "Moderate" : "green", "High": "orange"}
for level, subset in df.groupby("activity_level"):
    plt.scatter(subset["steps"], subset["calories_burned"],
                label = level, color=colors[level], alpha=0.6)
plt.title("Steps vs Calories-By Activity Level")
plt.xlabel("Steps")
plt.ylabel("Calories Burned")
plt.legend()
```

```python
plt.hist(df["steps"], bins=20, color="indigo", edgecolor="black")
plt.title("Distribution of Daily Steps")
plt.xlabel("Steps")
plt.ylabel("Frequency")
plt.show() #this eliminates the array being generated in addition to the plot
```

```python
fig,axes=plt.subplots(1,2)
#above says we want a figure with 1 row and 2 charts
axes[0].hist(df["sleep_hours"], bins=15, color="teal", edgecolor="black")
axes[0].set_title("Sleep Hours Distribution")
#we use set_title instead of just title when working with multiple axes

axes[1].hist(df["resting_hr"], bins=15, color= "darkred", edgecolor="black")
axes[1].set_title("Resting Heart Rate Distribution (bpm)")
```

```python
fig, axes = plt.subplots(2,2, figsize=(12,10))

# Histogram 1 (top left)
axes[0,0].hist(df["steps"], bins = 20, color="skyblue", edgecolor="black")
axes[0,0].set_title("Daily Steps")

# Histogram 2 (top right)
axes[0,1].hist(df["calories_burned"], bins=20, color="orange", edgecolor="black")
axes[0,1].set_title("Calories Burned")

# Histogram 3 (bottom left)
axes[1,0].hist(df["sleep_hours"], bins = 20, color="green", edgecolor="black")
axes[1,0].set_title("Number of Hours of Sleep")

#Histogram 4 (bottom right)
axes[1,1].hist(df["resting_hr"], bins = 20, color="darkred", edgecolor="black")
axes[1,1].set_title("Resting Heart Rate")

plt.tight_layout()
plt.show()
```

```python
activity_counts=df["activity_level"].value_counts()
plt.bar(activity_counts.index, activity_counts.values, color="pink")
plt.title("Activity Level Counts")
plt.xlabel("Activity Level")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.boxplot(column="sleep_hours", by="activity_level")
plt.title("Sleep Hours By Activity Level")
plt.suptitle("")
plt.xlabel("Activity Level")
plt.ylabel("Number of Hours of Sleep")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df.boxplot(column="resting_hr", by="activity_level")
plt.title("Resting Heart Rate by Activity Level")
plt.suptitle("")
plt.xlabel("Activity Level")
plt.ylabel("Resting Heart Rate")
plt.show()
```

```python
#rows of df where stress score is greater than 60

high_stress= df[df["stress_score"] > 60]
high_stress.head(10)
```

```python
plt.scatter(high_stress["sleep_hours"],high_stress["resting_hr"], color="deeppink")
plt.title("Sleep vs Resting Heart Rate High Stress Days")
plt.xlabel("Sleep Hours")
plt.ylabel("Resting Heart Rate")
plt.show()
```

```python
fig, axes = plt.subplots(1,3, figsize=(15, 5))

levels=["Low", "Moderate", "High"]
colors = ["skyblue", "limegreen", "orange"]

for ax, level, color in zip(axes, levels, colors): 
    subset = df[df["activity_level"] == level]
    ax.scatter(subset["steps"], subset["calories_burned"], alpha = 0.6, color =color)
    ax.set_title(f"{level} Activity")
    ax.set_xlabel("Steps")
    ax.set_ylabel("Calories Burned")

plt.tight_layout()
plt.show()
```

```python
plt.hist(df["resting_hr"], bins=20, alpha=0.5, label="Resting HR", color="cornflowerblue")
plt.hist(df["sleep_hours"], bins=20, alpha=0.5, label="Sleep Hours", color="palevioletred")
plt.title("Overlayed Distributions")
plt.legend()
```

```python
low_sleep = df[df["sleep_hours"] <= 6]
high_sleep = df[df["sleep_hours"] > 6]
plt.hist(low_sleep["resting_hr"], bins=20, alpha=0.5, label="Low Sleep", color="deeppink")
plt.hist(high_sleep["resting_hr"], bins=20, alpha=0.5, label="High Sleep", color= "palevioletred")
plt.title("Overlayed Distributions")
plt.show()
```
