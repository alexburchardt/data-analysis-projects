Importing applications and csvs

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
customers=pd.read_csv("customers.csv")
orders=pd.read_csv("orders.csv")
order_items=pd.read_csv("order_items.csv")
```

Sort the order date descending order

Filter the order items dataset to only show rows where quantity of items is 2 or higher. AND sort by quantity.

```python
over_two=order_items[order_items["quantity"]>=2].sort_values("quantity")
print(over_two)
```

Clean up the first & last name columns (remove leading spaces and capitalize them all)

```python
customers["first_name"]=customers["first_name"].str.strip().str.capitalize()
customers["last_name"]=customers["last_name"].str.strip().str.capitalize()
```

```python
customers.sort_values("last_name")
```

Look up all the different values in a given column--in this case, the city column, so we can clean up.

```python
customers["city"].value_counts()

#OR

customers["city"].unique()
```

Create a dictionary of values to be replaced/what to replace them with. 

```python
city_standardization= {'LA' : 'Los Angeles', 
                       'NYC' : 'New York'}
customers["city"]=customers["city"].replace(city_standardization)
customers["city"]=customers["city"].str.title()

customers["city"].value_counts()
```

Using .apply()

```python
customers["last_name"].value_counts()
#shows us some last names have space before them
```

When we use lambda, it creates a quick "inline" function. Defines a function really quickly, in this case we called it function 'x'. 'x' stands for the value and we define the function on the last line. With each last name being 'x', apply this function. In this case, apply the strip function.

```python
customers["clean_last_name"]= customers["last_name"].apply(
    lambda x:
    x.strip()
)
customers["clean_last_name"].value_counts()
```

Create a new column called total price (which is calculated by multiplying unit price by quantity)

```python
order_items["total_price"] = order_items["quantity"]*order_items["unit_price"]
order_items
```

Create new column called total_price_apply. Use .apply on each row (so to entire dataset). This does the same things as the coding above! Just another way to do it.

```python
order_items["total_price_apply"] = order_items.apply(
    lambda row:
    row["quantity"]*row["unit_price"], axis=1
)
order_items
```

Create a new column called order type. Use apply method on the order_items dataframe. Go through each row and run the code on it. If row's quantity is more than/equal to 3, fill the column with string "Large Expensive Order", and so forth.

```python
order_items["order_type"] = order_items.apply(
    lambda row: (
        "Large Expensive Order" if row ["quantity"] >= 3 and row["unit_price"] > 50
        else "Bulk Order" if row ["quantity"] >=3
        else "Standard Order"
    ), axis=1
)
order_items
```

Merging tables

```python
orders_with_customers=orders.merge(customers,how="inner")
orders_with_customers.head()
```

```python
orders_customers_items=orders_with_customers.merge(order_items, on="order_id", how="inner")
orders_customers_items.head()
```

```python
combined= (
    orders.merge(customers, on="customer_id", how="inner").merge(order_items, on="order_id", how="inner")
)
combined
```

Grouping and aggregation

```python
order_items["total_price"].sum()
```

```python
order_items.groupby("product_name")["total_price"].sum()
```

Write a line of code taht assesses the total quanity of units sold of each item, presented in descending order.

```python
order_items.groupby("product_name")["quantity"].sum().sort_values(ascending=False)
```

Use the agg method to group by product name and calculate the total revenue and average quantity bought. Agg is really flexible.

```python
order_items.groupby("product_name").agg(
    total_revenue=("total_price", "sum"),
    avg_quantity=("quantity", "mean")
)
```

```python
combined.groupby("customer_id")["total_price"].sum().sort_values(ascending=False).head()
```

```python
sample = combined[["customer_id", "product_name", "total_price"]].head(12)

pivot=sample.pivot_table(index="customer_id", columns="product_name", values="total_price", aggfunc="sum")

pivot
```
